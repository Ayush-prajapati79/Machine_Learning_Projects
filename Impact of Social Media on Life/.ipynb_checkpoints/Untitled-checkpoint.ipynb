```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select numerical columns
num_df = df.select_dtypes(include='number')

# Calculate correlation matrix
corr_matrix = num_df.corr()

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Matrix')
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x='Sleep_Duration_Hours',
    y='Daily_Usage_Hours',
    scatter_kws={'alpha': 0.4},
    line_kws={'color': 'red'}
)

plt.title('Screen Time vs Sleep Duration')
plt.xlabel('Daily Screen Time (Hours)')
plt.ylabel('Sleep Duration (Hours)')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.regplot(
    data=df,
    x='Daily_Usage_Hours',
    y='Sleep_Duration_Hours',
    scatter_kws={'alpha': 0.4},
    line_kws={'color': 'red'}
)

plt.title('Screen Time vs Sleep Duration')
plt.show()
```

```python
avg_sleep = df.groupby(
    'Daily_Usage_Hours'
)['Sleep_Duration_Hours'].mean().reset_index()

plt.figure(figsize=(7, 5))

sns.lineplot(
    data=avg_sleep,
    x='Daily_Usage_Hours',
    y='Sleep_Duration_Hours',
    marker='o'
)

plt.title('Average Sleep Duration by Screen Time')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.hexbin(
    df['Daily_Usage_Hours'],
    df['Sleep_Duration_Hours'],
    gridsize=20,
    cmap='Blues'
)

plt.colorbar(label='Number of Students')
plt.xlabel('Daily Screen Time (Hours)')
plt.ylabel('Sleep Duration (Hours)')
plt.title('Screen Time vs Sleep Density')
plt.show()
```

```python
df['Screen_Time_Group'] = pd.cut(
    df['Daily_Usage_Hours'],
    bins=[0, 2, 4, 6, 8, 24],
    labels=['0-2', '2-4', '4-6', '6-8', '8+']
)

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=df,
    x='Screen_Time_Group',
    y='Sleep_Duration_Hours'
)

plt.xlabel('Daily Screen Time (Hours)')
plt.ylabel('Sleep Duration (Hours)')
plt.title('Sleep Distribution by Screen Time')
plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(
    data=df,
    x='Screen_Time_Group',
    y='Sleep_Duration_Hours',
    errorbar=None
)

plt.xlabel('Daily Screen Time (Hours)')
plt.ylabel('Average Sleep Duration (Hours)')
plt.title('Average Sleep by Screen-Time Group')
plt.show()
```
